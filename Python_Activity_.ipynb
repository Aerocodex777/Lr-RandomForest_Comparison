```python
def factorial(n):
    result = 1
    i = 1
    while i <= n:
        result = result * i
        i += 1
    return result

print(factorial(5))
```

## Q4. Create a function that checks whether the given number is prime.

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

print(is_prime(17))
```

## Q5. Write a function to determine whether a given number is a palindrome.

```python
def number_palindrome(n):
    original = n
    reverse = 0
    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    return original == reverse

print(number_palindrome(121))
```

## Q9. Write a function to check whether a given number is an Armstrong number.

```python
def armstrong(n):
    if n < 0:
        return False

    original = n
    temp = n
    digits = 0

    if temp == 0:
        digits = 1
    else:
        while temp > 0:
            digits += 1
            temp = temp // 10

    temp = n
    total = 0

    while temp > 0:
        digit = temp % 10
        power = 1
        i = 0
        while i < digits:
            power = power * digit
            i += 1
        total += power
        temp = temp // 10

    return total == original

print(armstrong(153))
```

## Q10. Create a function to generate the Fibonacci sequence up to a specified number of terms.

```python
def fibonacci(n):
    a = 0
    b = 1
    result = []
    i = 0

    while i < n:
        result.append(a)
        a, b = b, a + b
        i += 1

    return result

print(fibonacci(10))
```

## Q17. Write a function to identify the duplicate elements present in a list.

```python
def duplicates(numbers):
    result = []
    i = 0

    while i < len(numbers):
        j = i + 1
        found = False

        while j < len(numbers):
            if numbers[i] == numbers[j]:
                found = True
            j += 1

        if found:
            already_added = False
            k = 0
            while k < len(result):
                if result[k] == numbers[i]:
                    already_added = True
                k += 1
            if not already_added:
                result.append(numbers[i])

        i += 1

    return result

print(duplicates([1, 2, 3, 2, 4, 5, 1, 3]))
```

## Q23. Write a function that counts the number of vowels and consonants in a string.

## Q27. Write a function to determine whether two given strings are anagrams.

```python
def are_anagrams(a, b):
    if len(a) != len(b):
        return False

    freq_a = {}
    freq_b = {}

    i = 0
    while i < len(a):
        ch = a[i]
        if ch in freq_a:
            freq_a[ch] += 1
        else:
            freq_a[ch] = 1
        i += 1

    i = 0
    while i < len(b):
        ch = b[i]
        if ch in freq_b:
            freq_b[ch] += 1
        else:
            freq_b[ch] = 1
        i += 1

    return freq_a == freq_b

print(are_anagrams("listen", "silent"))
```

## Q28. Create a function that finds the longest word in a given sentence.

```python
def longest_word(sentence):
    longest = ""
    current = ""
    i = 0

    while i < len(sentence):
        if sentence[i] != ' ':
            current += sentence[i]
        else:
            if len(current) > len(longest):
                longest = current
            current = ""
        i += 1

    if len(current) > len(longest):
        longest = current

    return longest

print(longest_word("Python is a powerful programming language"))
```

## Q44. Create a function to flatten a nested list into a single-level list.

```python
def flatten_list(nested):
    result = []

    def flatten(items):
        i = 0
        while i < len(items):
            if isinstance(items[i], list):
                flatten(items[i])
            else:
                result.append(items[i])
            i += 1

    flatten(nested)
    return result

print(flatten_list([1, [2, 3], [4, [5, 6]]]))
```

## Q45. Write a function to determine the maximum possible sum of a contiguous subarray.

```python
def max_subarray_sum(numbers):
    current = numbers[0]
    best = numbers[0]
    i = 1

    while i < len(numbers):
        if current + numbers[i] > numbers[i]:
            current = current + numbers[i]
        else:
            current = numbers[i]

        if current > best:
            best = current

        i += 1

    return best

print(max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

## Q46. Create a function that rotates the elements of a list by k positions.

```python
def rotate_list(numbers, k):
    n = len(numbers)

    if n == 0:
        return numbers

    k = k % n
    result = []

    i = n - k
    while i < n:
        result.append(numbers[i])
        i += 1

    i = 0
    while i < n - k:
        result.append(numbers[i])
        i += 1

    return result

print(rotate_list([1, 2, 3, 4, 5], 2))
```

## Q47. Write a function to find the nth prime number.

```python
def nth_prime(n):
    count = 0
    number = 1

    while count < n:
        number += 1
        prime = True
        divisor = 2

        while divisor * divisor <= number:
            if number % divisor == 0:
                prime = False
                break
            divisor += 1

        if prime:
            count += 1

    return number

print(nth_prime(10))
```

## Q48. Create a function to calculate the Greatest Common Divisor (GCD) of two numbers.

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print(gcd(48, 18))
```

## Q49. Create a function to calculate the Least Common Multiple (LCM) of two numbers.

```python
def lcm(a, b):
    x = a
    y = b

    while y != 0:
        x, y = y, x % y

    gcd_value = x
    return (a * b) // gcd_value

print(lcm(12, 18))
```

## Q50. Create a function to find the longest consecutive sequence of numbers in a given list.

```python
def longest_consecutive(numbers):
    if len(numbers) == 0:
        return 0

    longest = 0
    i = 0

    while i < len(numbers):
        start = numbers[i]

        # Check whether start-1 exists.
        has_previous = False
        j = 0
        while j < len(numbers):
            if numbers[j] == start - 1:
                has_previous = True
                break
            j += 1

        if not has_previous:
            length = 1
            current = start + 1

            while True:
                found = False
                j = 0
                while j < len(numbers):
                    if numbers[j] == current:
                        found = True
                        break
                    j += 1

                if not found:
                    break

                length += 1
                current += 1

            if length > longest:
                longest = length

        i += 1

    return longest

print(longest_consecutive([100, 4, 200, 1, 3, 2]))
```

## Q9. Abstract Employee with abstract calculate_salary(); FullTimeEmployee and PartTimeEmployee implement salary calculation.

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, monthly_salary):
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        return self.monthly_salary

class PartTimeEmployee(Employee):
    def __init__(self, hours, rate):
        self.hours = hours
        self.rate = rate

    def calculate_salary(self):
        return self.hours * self.rate

full_time = FullTimeEmployee(50000)
part_time = PartTimeEmployee(80, 400)

print("Full-time salary:", full_time.calculate_salary())
print("Part-time salary:", part_time.calculate_salary())
```
